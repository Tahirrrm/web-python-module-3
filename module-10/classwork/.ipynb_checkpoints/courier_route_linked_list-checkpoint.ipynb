### ЗАДАЧА: Маршрут курьера на связанном списке

Логист получает маршрут курьера как цепочку точек доставки.
Во время дня маршрут может меняться:
- в конец добавляются новые точки,
- после конкретной точки может появиться срочная доставка,
- первая точка может быть уже выполнена и удалена из маршрута.

Нужно собрать модель маршрута, которая:
- хранит точки в правильном порядке,
- умеет добавлять новую точку в конец,
- умеет вставлять точку после указанной,
- умеет удалять уже выполненную первую точку,
- в конце показывает весь оставшийся маршрут.


НЕОБХОДИМО РЕАЛИЗОВАТЬ:

1. Класс `StopNode` с полями `address`, `next`.

2. Класс `CourierRoute`, который:
   - хранит голову маршрута,
   - умеет добавлять точку в конец через `append_stop(address)`,
   - умеет вставлять точку после указанной через `insert_after(target, address)`,
   - умеет удалять первую точку через `pop_first_stop()`,
   - умеет возвращать маршрут списком через `to_list()`.

3. Выполнить команды из списка `actions`.

4. Вывести:
   - маршрут до изменений,
   - результаты операций,
   - финальный маршрут.


```python
from dataclasses import dataclass
from typing import Optional


initial_stops = [
    "Склад",
    "ул. Лесная, 10",
    "пр. Мира, 25",
    "наб. Реки, 7",
]

actions = [
    ("append", "ул. Садовая, 3"),
    ("insert_after", "ул. Лесная, 10", "б-р Центральный, 8"),
    ("pop_first", None),
    ("insert_after", "несуществующий адрес", "ул. Тихая, 11"),
    ("append", "ул. Новая, 15"),
]


@dataclass
class StopNode:
    address: str
    next: Optional["StopNode"] = None


class CourierRoute:
    def __init__(self):
        # TODO 1: завести поле head
        # TODO 2: в пустом маршруте head должен быть None
        self.head= None
 

    def append_stop(self, address: str) -> None:
        # TODO 1: создать новый узел StopNode(address)
        new_node = StopNode(address)
        # TODO 2: если head пустой, сделать новый узел головой и завершить метод
        if self.head is None:
        # TODO 3: иначе пройти по списку до последнего узла (у которого next == None)
            self.head=new_node
        else:   
        # TODO 4: привязать новый узел в конец: last.next = new_node
            current = self.head
            while current.next is not None:
                current = current.next
            current.next = new_node
        

    def insert_after(self, target: str, address: str) -> bool:
        # TODO 1: начать обход с head
        current = self.head
        # TODO 2: пока текущий узел существует:
        while current is not None:
        #   - если current.address == target:
            if current.address == target:
        #       1) создать новый узел
                new_node = StopNode(address)
        #       2) новый узел должен ссылаться на current.next
                new_node.next = current.next
        #       3) current.next перенаправить на новый узел
                current.next = new_node
        #       4) вернуть True
                return True
        #   - иначе перейти к следующему узлу
            current = current.next
        # TODO 3: если дошли до конца и target не найден, вернуть False
        return False
                

    def pop_first_stop(self) -> Optional[str]:
        # TODO 1: если head == None, вернуть None
        if self.head is None:
            return None
        # TODO 2: сохранить адрес из head.address
        removed_address= self.head.address
        # TODO 3: сдвинуть голову на следующий узел: head = head.next
        self.head = self.head.next
        # TODO 4: вернуть сохраненный адрес
        return removed_address
     
    def to_list(self) -> list[str]:
        # TODO 1: создать пустой список result
        result= []
        # TODO 2: пройтись от head до конца списка
        current = self.head
        # TODO 3: на каждом узле добавлять current.address в result
        while current is not None:
            result.append(current.address)
            current=current.next
        # TODO 4: вернуть result
        return result


route = CourierRoute()
for stop in initial_stops:
    route.append_stop(stop)

print("Стартовый маршрут:", route.to_list())

for command, value1, *rest in [(a[0], a[1], None) if a[0] != "insert_after" else (a[0], a[1], a[2]) for a in actions]:
    if command == "append":
        route.append_stop(value1)
        print(f"append ->", {value1})
    elif command == "insert_after":
        target = value1
        new_address = rest[0]
        print(f"insert_after ->", {target}, {route.insert_after(target, new_address)})
    elif command == "pop_first":
        print(f"pop_first ->", {route.pop_first_stop()})

print(f"Финальный маршрут:", {route.to_list()})
```
