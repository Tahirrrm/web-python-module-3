## Учёт заявок в IT-поддержке”

В компании есть небольшая IT-поддержка. Сотрудники отправляют заявки: `не работает принтер`, `забыл пароль`, `нет интернета`. Нужно сделать консольную программу, которая помогает хранить и обрабатывать эти заявки.

### Что должна уметь программа

Меню:
1. Добавить заявку
2. Показать все заявки
3. Изменить статус заявки
4. Показать открытые заявки
5. Найти заявку по имени сотрудника
0. Выход

### Данные одной заявки
```txt
Номер заявки
Имя сотрудника
Описание проблемы
Статус
```

**Статусы можно сделать такими:**
`новая`, `в работе`, `закрыта`


### Пример заявки
```txt
№1
Сотрудник: Иван
Проблема: Не работает принтер
Статус: новая
```

### Обязательные условия
1. Номер заявки создаётся автоматически.
2. При добавлении заявки статус всегда `новая`.
3. Можно изменить статус только на:
    - `новая`
    - `в работе`
    - `закрыта`
4. При поиске по имени программа должна показывать все заявки этого сотрудника.
5. Открытые заявки — это заявки со статусом `новая` или `в работе`.
6. Решение должно быть реализовано в объектно-ориентированном стиле с соблюдением принципов `SOLID`:
   - отдельный класс для заявки;
   - отдельный класс или сервис для управления заявками;
   - отдельный класс для работы с консольным меню;
   - классы не должны брать на себя лишние обязанности;
   - зависимости между частями программы должны быть минимальными и понятными.

````python
from dataclasses import dataclass
from typing import List

@dataclass
class Ticket:
    id : int
    employee_name: str
    description: str
    status: str = "новая"


    def __str__(self) -> str:
        return f"№{self.id}\nСотрудник: {self.employee_name}\nПроблема: {self.description}\nСтатус: {self.status}"
    
class TicketMemory:

    def __init__(self):
        self._tickets: List[Ticket] = []
        self._next_id = 1

    def add_ticket(self,employee_name: str, description: str) -> Ticket:
        ticket = Ticket(id=self._next_id,employee_name=employee_name,description=description,status ="новая")
        self._tickets.append(ticket)
        self._next_id += 1
        return ticket

    def get_all_tickets(self) -> List[Ticket]:
        return self._tickets.copy()
    
    def update_ticket_status(self,ticked_id: int, new_status: str) -> bool:
        for ticket in self._tickets:
            if ticket.id == ticked_id:
                ticket.status = new_status
                return True
        return False

    def find_tickets_by_employee(self,employee_name: str) -> List [Ticket]:
        return [
            ticket for ticket in self._tickets
            if employee_name.lower() in ticket.employee_name.lower()
        ]
    
class TicketService:

    ValidStatuses = ["новая","в работе","закрыта"]

    def __init__(self,repository: TicketMemory):
        self.repository = repository

    def create_ticket (self, employee_name :str, description: str) -> Ticket:
        return self.repository.add_ticket(employee_name,description)
    
    def get_all_tickets(self)-> List [Ticket]:
        return self.repository.get_all_tickets()
    
    def update_ticket_status(self, ticket_id: int, new_status: str ) -> bool:
        if new_status not in self.ValidStatuses:
            return False
        return self.repository.update_ticket_status(ticket_id, new_status)
    
    def get_open_tickets(self) -> List[Ticket]:
        all_tickets = self.get_all_tickets()
        return [ticket for ticket in all_tickets if ticket.status in  ["новая", "в работе"] ]
    
    def find_tickets_by_employee( self, employee_name: str) -> List [Ticket]:
        return self.repository.find_tickets_by_employee(employee_name)
    
class ConsoleMenu:

    def __init__(self,ticket_service: TicketService):
        self.ticket_service = ticket_service

    def display(self):

        while True:
            print("\n=== Учёт заявок в IT‑поддержке ===")
            print("1. Добавить заявку")
            print("2. Показать все заявки")
            print("3. Изменить статус заявки")
            print("4. Показать открытые заявки")
            print("5. Найти заявку по имени сотрудника")
            print("0. Выход")

            choice = input("Выберите действие: ").strip()

            if choice == "1":
                self._add_ticket()
            elif choice == "2":
                self._show_all_tickets()
            elif choice == "3":
                self._update_ticket_status()
            elif choice == "4":
                self._show_open_tickets()
            elif choice == "5":
                self._find_tickets_by_employee()
            elif choice == "0":
                print("Выход из программы.")
                break
            else:
                print("Неверный выбор. Попробуйте снова.")


    def _add_ticket(self):
        print( "\n --Добавление новой заявки ---")
        employee_name = input ("Имя сотрудника: ").strip()
        description = input( "Описание проблемы:").strip()

        if not employee_name or not description:
            print("Ошибка: имя сотрудника и описание проблемы обязательны.")
            return
        
        ticket = self.ticket_service.create_ticket(employee_name, description)
        print(f"Заявка №{ticket.id} успешно добавлена!")


    def _show_all_tickets(self):
        tickets = self.ticket_service.get_all_tickets()
        if not tickets:
            print("Заявок нет.")
        else:
            for ticket in tickets:
                print(f"\n{ticket}")


    def _update_ticket_status(self):
        try:
            ticket_id = int(input("Введите номер заявки: "))
            print("Доступные статусы: новая, в работе, закрыта")
            new_status = input("Новый статус: ").strip()

            if self.ticket_service.update_ticket_status(ticket_id, new_status):
                print(f"Статус заявки №{ticket_id} обновлён.")
            else:
                print("Заявка не найдена или неверный статус.")
        except ValueError:
            print("Ошибка: номер заявки должен быть числом.")


    def _show_open_tickets(self):
        open_tickets = self.ticket_service.get_open_tickets()
        if not open_tickets:
            print("Открытых заявок нет.")
        else:
            for ticket in open_tickets:
                print(f"\n{ticket}")


    def _find_tickets_by_employee(self):
        employee_name = input("Введите имя сотрудника: ").strip()
        if not employee_name:
            print("Ошибка: имя сотрудника не может быть пустым.")
            return

        found_tickets = self.ticket_service.find_tickets_by_employee(employee_name)
        if not found_tickets:
            print(f"Заявок от сотрудника '{employee_name}' не найдено.")
        else:
            for ticket in found_tickets:
                print(f"\n{ticket}")
def main():

    repository = TicketMemory()
    ticket_service = TicketService(repository)
    menu = ConsoleMenu(ticket_service)
    menu.display()

if __name__ == "__main__":
    main()
````
